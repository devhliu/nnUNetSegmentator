## Basic Usage

### Single Image Segmentation

Here's how to segment a single medical image:

```python
from nnunetsegmentator import SegmentationOrchestrator, Config

# Initialize orchestrator with a task
orchestrator = SegmentationOrchestrator(
    task_name='gtrc',
    config=Config(use_gpu=True, num_workers=4)
)

# Segment a single image
result = orchestrator.segment(
    input_data='path/to/your/image.nii.gz',
    output_path='path/to/output/segmentation.nii.gz',
    return_labels=True,
    compute_metrics=True
)

# Print results
print(f"Segmentation shape: {result.segmentation.GetSize()}")
print(f"Labels detected: {list(result.labels.keys())}")

if result.metrics:
    print(f"Volume: {result.metrics['volume_mm3']:.2f} mm³")
```

### Available Tasks

nnunetsegmentator supports multiple pre-configured tasks:

```python
from nnunetsegmentator import TaskRegistry

# List all available tasks
tasks = TaskRegistry.list_tasks()
print("Available tasks:")
for task in tasks:
    print(f"  - {task}")
```

## Advanced Usage

### Custom Configuration

You can customize the segmentation process with various options:

```python
from nnunetsegmentator import Config

# Create custom configuration
config = Config(
    use_gpu=True,
    num_workers=4,
    batch_size=4,
    patch_size=[128, 128, 64],
    min_size=100,
    confidence_threshold=0.5
)

# Use custom config
orchestrator = SegmentationOrchestrator(
    task_name='gtrc',
    config=config
)
```

### Post-processing Options

Apply post-processing to refine segmentation results:

```python
result = orchestrator.segment(
    input_data='path/to/image.nii.gz',
    output_path='path/to/output.nii.gz',
    return_labels=True,
    compute_metrics=True,
    postprocess=True,
    fill_holes=True,
    remove_small_components=True
)
```
